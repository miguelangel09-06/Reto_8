A continuacion esas 3 funciones con lambda=

```python
#importamos esto por que vamos a usar operaciones matematicas
import math
math.pi#esto para el número pi
radio_e=(float(input("Por favor ingrese el radio de la esfera en cm:")))
conito_r=(float(input("Por favor ingrese el radio del cono en cm:")))
conito_a=(float(input("Por favor ingrese la altura del cono en cm:")))

#lambda para determinar el volumen de la esfera

f_volumen=(lambda x:(4/3)*math.pi*x**3)
volumen_esfera=f_volumen(radio_e)
print("El volumen de una esfera de ",radio_e, "cm de radio es= ",volumen_esfera, " cm^3")

#lambda para determinar el area superficial de la esfera
f_superficial=(lambda y:(4)*math.pi*y**2)
area_superficial=f_superficial(radio_e)
print("El area superficial de una esfera de ",radio_e, "cm de radio es= ",area_superficial, " cm^2")

#lambda para determinar el volumen de un cono

f_volumen_c=(lambda r,a:(1/3)*math.pi*r**2*a)
volumen_de_cono=f_volumen_c(conito_r,conito_a)
print("El volumen de un cono de",conito_r, " cm de radio y una altura de",conito_a," cm es:",volumen_de_cono," cm^2")


```

3.Escriba una función recursiva para calcular la operación de la potencia.

```python
#La funcion toma 2 argumentos que son la base y el exponente
def operacion_potencia(b,e)->int:
    if e==0: #Si es 0 retorna 1 
        return 1
    else:
        return b*operacion_potencia(b,e-1) #Operacion recursiva 
#Se piden los datos
base=int(input("Por favor ingrese la base de la potencia:"))
exponente=int(input("Por favor ingrese el exponente:"))
#Se imprime 
print("El resultado de ",base," elevado a la ",exponente,"es:",operacion_potencia(base,exponente))
```
